## 1. Variables & Rules

Variabel adalah kontainer yang digunakan untuk menyimpan nilai data di memori komputer. Setiap variabel memiliki nama unik yang merujuk pada alamat memori nilai tersebut.

### Aturan Penamaan Variabel
- **Karakter yang diizinkan**: Huruf (`a-z`, `A-Z`), angka (`0-9`), dan underscore (`_`).
- **Tidak Boleh Diawali Angka**: Harus diawali huruf atau underscore (contoh: `variable1` valid, `1variable` invalid).
- **Case Sensitive**: Variabel `a` dan `A` dianggap berbeda oleh Python.
- **Gaya Penulisan**: `snake_case` (direkomendasikan Python), `camelCase`, `PascalCase`.

```python
# Menyimpan hasil kalkulasi ke variabel 'a'
a = 5 + 6

# Memanggil nilai variabel 'a'
print("Nilai variabel a:", a) # Output: 11

# Contoh variabel dengan konteks & readability
total_price = 78000
discount_rate = 0.25
amount_disc = total_price * discount_rate
prince_after_disc = total_price - amount_disc

print("Besar Diskon:", amount_disc)
print("Harga Setelah Diskon:", prince_after_disc)
```

## 2. Built-in Data Types & Duck Typing

### Hirarki Tipe Data Python
- **Primitive Data Types**: Integer, Float, String, Boolean, Complex.
- **Non-Primitive Data Types**: List, Tuple, Set, Dictionary.

### Filosofi Duck Typing
"*If it walks like a duck and quacks like a duck, it's a duck.*"
Python secara otomatis mengenali tipe data berdasarkan nilai yang diisikan ke variabel tanpa deklarasi eksplisit.

```python
# Python otomatis mendeteksi tipe data (Duck Typing)
age = 25           # int
price = 49.99      # float
name = "DigitalSkola" # str
is_active = True   # bool

print(f"age: {age} -> {type(age)}")
print(f"price: {price} -> {type(price)}")
print(f"name: {name} -> {type(name)}")
print(f"is_active: {is_active} -> {type(is_active)}")
```

## 3. Deep Into Collections

### A. List (Ordered, Mutable, Allow Duplicates)
Digunakan untuk menyimpan sekumpulan data terurut yang dapat diubah.

```python
# Akses & Konversi List
a = [10, 20, 30]
print("Elemen pertama:", a[0])
print("Elemen terakhir:", a[-1])

# Differences: Append vs Extend
list_a = [1, 2]
list_a.append([3, 4])
print("Result append:", list_a) # Result: [1, 2, [3, 4]]

list_b = [1, 2]
list_b.extend([3, 4])
print("Result extend:", list_b) # Result: [1, 2, 3, 4]
```

### B. Tuple (Ordered, Immutable, Allow Duplicates)
Koleksi data terurut yang **tidak dapat diubah** setelah diinisialisasi.

```python
# Inisialisasi & Akses Tuple
tp = ("ID", 101, True)
print("Indeks 0:", tp[0])

# Konversi List ke Tuple
converted_tuple = tuple([1, 2, 3])
print("Tuple hasil konversi:", converted_tuple)
```

### C. Set (Unordered, Unique Values Only)
Koleksi tak terurut yang secara otomatis **mengeliminasi nilai duplikat**.

```python
# Eliminasi Duplikat otomatis
st = {1, 2, 2, 3}
print("Set unik:", st) # Output: {1, 2, 3}

# Update dengan List
st.update([4, 5])
print("Set setelah update list:", st)

# Update dengan String (String diurai menjadi karakter unik)
st_str = {1, 2}
st_str.update('halo')
print("Set setelah update string:", st_str)
```

### D. Dictionary (Key-Value Pair, Immutable Key, Mutable Value)
Menyimpan data dalam pasangan kata kunci (key) dan nilai (value) untuk *lookup* data secara cepat.

```python
# Membuat & Akses Dictionary
student = {'id': 101, 'name': 'Budi'}
print("Nama siswa:", student['name'])

# Update Data
student['grade'] = 'A'
student.update({'id': 102})
print("Dictionary terupdate:", student)

# Konversi Pasangan Tuple ke Dict
dict_data = dict([
    ('role', 'admin'),
    ('level', 5)
])
print("Dict dari tuple:", dict_data)
```

## 4. Control Flow & Loops

Control Flow mengatur urutan dan kondisi baris kode yang dieksekusi (*conditional statement* dan *looping*).

```python
# Conditional Statement (IF - ELIF - ELSE)
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
else:
    print("Grade: C")

# For Loop dengan enumerate
items = ['A', 'B', 'C']
print("\n--- Iterasi For Loop ---")
for idx, x in enumerate(items):
    print(f"Index {idx}: {x}")

# While Loop
print("\n--- Iterasi While Loop ---")
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
```
